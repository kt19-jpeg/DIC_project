```python
raw_df = spark.read.csv(
    "/Volumes/workspace/default/dic_project/cleaned_drug_overdose_deaths.csv",
    header=True,
    inferSchema=True
)

display(raw_df)
```

```python
raw_df = raw_df.toDF(*[c.replace(" ", "_") for c in raw_df.columns])

print("Renamed columns:", raw_df.columns)
```

```python
raw_df.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_overdose")

print("Bronze table created successfully")
print(f"Total rows: {raw_df.count()}")
print(f"Columns: {raw_df.columns}")
```
